```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

cic = pd.read_csv("../data/processed/cicids2017_wednesday_clean.csv")
unsw_train_raw = pd.read_csv("../data/raw/unsw-nb15/UNSW_NB15_training-set.csv")
y_unsw_multiclass = unsw_train_raw['attack_cat']
```

```python
# Visualize class balance(Benign vs Attack)

plt.figure(figsize=(6,4))
sns.countplot(x='Label', data=cic, order=cic['Label'].value_counts().index)
plt.xticks(rotation=45, ha='right')
plt.title("CICIDS2017 (Wednesday) - Traffic Class Distribution")
plt.tight_layout()
plt.show()
```

```python
# Compare a key feature's distribution: Benign vs Attack
plt.figure(figsize=(8,5))
sns.boxplot(x='Label_binary', y='Flow Duration', data=cic)
plt.title("Flow Duration: Benign(0) vs Attack(1)")
plt.yscale('log')
plt.show()
```

```python
# Correlation heatmap of features
sample_cols= ['Flow Duration', 'Total Fwd Packets', 'Total Backward Packets',
                'Flow Bytes/s', 'Flow Packets/s', 'Fwd Packet Length Mean',
                'Bwd Packet Length Mean', 'Label_binary']
plt.figure(figsize=(10, 8))
sns.heatmap(cic[sample_cols].corr(), annot=True, cmap='coolwarm', center=0)
plt.title("Feature Correlation(sample)")
plt.tight_layout()
plt.show()

```

```python
# Visualize class balance(Benign vs Attack) for attack_cat
plt.figure(figsize=(8,5))
sns.countplot(y=y_unsw_multiclass,order=y_unsw_multiclass.value_counts().index)
plt.title("UNSW-NB15 - Attack Category Distribution")
plt.tight_layout()
plt.show()
```
